```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentences = [
    "I love learning artificial intelligence",
    "Machine learning is a powerful technology",
    "Deep learning is a part of machine learning",
    "Artificial intelligence is changing the world",
    "Python is widely used for machine learning",
    "Java is a popular programming language",
    "I enjoy writing Python programs",
    "Data science involves statistics and programming",
    "Natural language processing works with human language",
    "Word embeddings represent words as vectors",
    "Machine learning models learn from data",
    "Deep learning uses neural networks",
    "Neural networks are inspired by the human brain",
    "Technology is developing very quickly",
    "Students learn programming through practice",
    "Programming requires logical thinking",
    "I practice coding every day",
    "Artificial intelligence has many applications",
    "Robotics combines software and hardware",
    "AI and robotics are interesting fields"
]
```

```python
##Define vocaublary size
voc_size=10000
```

```python
###one hot representaion for every word
one_hot_representaion=[one_hot(words,voc_size)for words in sentences]
one_hot_representaion
```

```python
## word embedding representation
from tensorflow.keras.layers import Embedding 
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length=8
embedded_docs=pad_sequences(one_hot_representaion,padding='pre',maxlen=sent_length)
print(embedded_docs)
```

```python
## feature representataion
dim=10
```

```python
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
